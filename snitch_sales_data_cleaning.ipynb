```python
import pandas as pd
```

```python
df = pd.read_csv('snitch_uncleaned.csv')
print(sorted(df['City'].unique()))
```

```python
df['City'] = df['City'].str.strip().str.title()

city_mapping = {
    'Bengaluru': 'Bangalore',
    'Banglore': 'Bangalore',   
    'Hyd': 'Hyderabad',
    'Hyderbad': 'Hyderabad'
}

df['City'] = df['City'].replace(city_mapping)

print(sorted(df['City'].unique()))
```

```python
print(df['Discount_%'].describe())
print()

over_100 = df[df['Discount_%']>1.0]
print("rows with discount > 100%:",len(over_100))
print(over_100['Discount_%'].describe())
```

```python
df.loc[df['Discount_%']>1.0,'Discount_%']=None
print(df['Discount_%'].describe())
print("Missing Discount_% ab:", df['Discount_%'].isnull().sum())
```

```python
avg_discount = df['Discount_%'].mean()
print('avg discount (before filling):',avg_discount)

df['Discount_%'] = df['Discount_%'].fillna(avg_discount)

print()
print(df['Discount_%'].describe())
print("Missing ab:", df['Discount_%'].isnull().sum())
```

```python
negative_units = df[df['Units_Sold']< 0]

print("negative units_sold rows:", len(negative_units))
print("profit starts for these rows")
print(negative_units['Profit'].describe())
print("how many of these negative profit too",(negative_units['Profit']<0).sum())
```

```python
df.loc[df['Units_Sold'] < 0,'Units_Sold']=None

print(df['Units_Sold'].describe())
print("Missing Units_Sold ab:", df['Units_Sold'].isnull().sum())
```

```python
avg_units = df['Units_Sold'].mean()
print("avg_units_sold(before filling):",avg_units)

df['Units_Sold']=df['Units_Sold'].fillna(avg_units)

print(df['Units_Sold'].describe())
print("Missing Units_Sold ab:", df['Units_Sold'].isnull().sum())
```

```python
df_fresh = pd.read_csv('snitch_uncleaned.csv')
df['Sales_Amount']=df_fresh['Sales_Amount']

negative_sales = df[df['Sales_Amount']< 0]
print("negative sales amount:", len(negative_sales))

df.loc[df['Sales_Amount']<0,'Sales_Amount']=None

avg_sales_amount = df['Sales_Amount'].mean()
print("avg_sales_amount(before filling)",avg_sales_amount)

df['Sales_Amount']= df['Sales_Amount'].fillna(avg_sales_amount)

print(df['Sales_Amount'].describe())
print("missing value_amount ab:",df['Sales_Amount'].isnull().sum())
```
